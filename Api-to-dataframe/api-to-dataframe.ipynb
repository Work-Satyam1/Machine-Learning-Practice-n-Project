```python
import requests
import pandas as pd

```

```python
url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url)
```

```python
data = response.json()

df = pd.DataFrame(data)

df.head()
```

```python
df = pd.json_normalize(data)



df.head()
```

```python
df.columns
```

```python
df=df[[
    "id",
    "name",
    "username",
    "email",
    "phone",
    "website",
    "address.city",
    "company.name"
]]
```

```python
df = df.rename(columns={
    "address.city": "city",
    "company.name": "company"
})
```
